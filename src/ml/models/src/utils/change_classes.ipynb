```python
from pathlib import Path
```

```python
BASE_DIR = Path('/Users/lukepitstick/Projects/Data-Science/InfraDrone/python/datasets/RD2022')
NEW_LABELS_DIR = Path('/Users/lukepitstick/Projects/Data-Science/InfraDrone/python/datasets/new_labels')
```

```python
CLASS_MAP = {
    "longitudinal crack": 0,
    "transverse crack": 1,
    "alligator crack": 2,
    "other corruption": 3,
    "pothole": 4,
}
```

```python
def change_class(label_line):
    parts = label_line.strip().split()
    if len(parts) != 5:
        return None

    class_id = int(parts[0])
    if class_id in [0, 1, 2]:
        parts[0] = "0"
    elif class_id == 4:
        parts[0] = "1"
    else:
        return None

    return " ".join(parts)
```

```python
for split in ['train', 'val', 'test']:
    source_labels_dir = BASE_DIR / split / 'labels'
    output_labels_dir = NEW_LABELS_DIR / split / 'labels'
    output_labels_dir.mkdir(parents=True, exist_ok=True)

    for label_file in source_labels_dir.glob('*.txt'):
        output_file = output_labels_dir / label_file.name

        with label_file.open('r') as source, output_file.open('w') as target:
            for line in source:
                new_label = change_class(line)
                if new_label is not None:
                    target.write(f'{new_label}\n')
```
