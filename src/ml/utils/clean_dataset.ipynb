```python
import os

from pathlib import Path


BASE_DIR = Path('/Users/lukepitstick/Projects/Data-Science/InfraDrone/datasets/RD2022/')
```

```python
for split in ['train', 'val', 'test']:
    source_labels_dir = BASE_DIR / split / 'images'
    
    print(source_labels_dir)
    
    print(f"Number of files in {split}: {len([f for f in source_labels_dir.rglob('*') if f.is_file()])}")

    for image_file in source_labels_dir.glob('*.jpg'):
        label_file = BASE_DIR / split / 'labels' / image_file.with_suffix('.txt').name
        
        if not label_file.exists():
            continue
        
        with open(label_file, 'r') as f:
            lines = f.read()
            # Delete empty label files and images with no labels
            if lines == "" or lines == "\n":
                print(f"Deleting empty label file: {label_file}")
                os.remove(image_file)
                os.remove(label_file)
                continue
```
