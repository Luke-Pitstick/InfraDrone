```python
from pathlib import Path
import random
import shutil


# Load Dataset
DATASET_PATH = Path("/Users/lukepitstick/Projects/Data-Science/InfraDrone/datasets/segmentation/crack_segmentation_dataset")
```

```python
# Count Test and Train Files
train_files = sorted(
    p for p in (DATASET_PATH / "train/images").iterdir()
    if p.is_file() and p.suffix.lower() in {".jpg", ".jpeg", ".png"}
)
test_files = sorted(
    p for p in (DATASET_PATH / "test/images").iterdir()
    if p.is_file() and p.suffix.lower() in {".jpg", ".jpeg", ".png"}
)

len_train_files = len(train_files)
len_test_files = len(test_files)

print(f"Train Files: {len_train_files}")
print(f"Test Files: {len_test_files}")
```

```python
# Create val dir with 20% of train files
VAL_RATIO = 0.2
RANDOM_SEED = 42

random.seed(RANDOM_SEED)
val_count = int(len_train_files * VAL_RATIO)
val_files = random.sample(train_files, val_count)

val_image_dir = DATASET_PATH / "val/images"
val_mask_dir = DATASET_PATH / "val/masks"
val_label_dir = DATASET_PATH / "val/labels"

for directory in (val_image_dir, val_mask_dir, val_label_dir):
    directory.mkdir(parents=True, exist_ok=True)

moved = 0
skipped = 0

for image_path in val_files:
    mask_path = DATASET_PATH / "train/masks" / image_path.name
    label_path = DATASET_PATH / "train/labels" / f"{image_path.stem}.txt"

    dest_image = val_image_dir / image_path.name
    dest_mask = val_mask_dir / mask_path.name
    dest_label = val_label_dir / label_path.name

    missing = [p for p in (image_path, mask_path, label_path) if not p.exists()]
    if missing:
        skipped += 1
        print(f"Skipping {image_path.name}: missing {[p.name for p in missing]}")
        continue

    shutil.move(image_path, dest_image)
    shutil.move(mask_path, dest_mask)
    shutil.move(label_path, dest_label)
    moved += 1

print(f"Moved {moved} samples to val")
print(f"Skipped {skipped} samples")
```

```python
train_files = sorted(
    p for p in (DATASET_PATH / "train/images").iterdir()
    if p.is_file() and p.suffix.lower() in {".jpg", ".jpeg", ".png"}
)
test_files = sorted(
    p for p in (DATASET_PATH / "test/images").iterdir()
    if p.is_file() and p.suffix.lower() in {".jpg", ".jpeg", ".png"}
)

val_files = sorted(
    p for p in (DATASET_PATH / "val/images").iterdir()
    if p.is_file() and p.suffix.lower() in {".jpg", ".jpeg", ".png"}
)

len_train_files = len(train_files)
len_test_files = len(test_files)
len_val_files = len(val_files)

print(f"Train Files: {len_train_files}")
print(f"Test Files: {len_test_files}")
print(f"Val Files: {len_val_files}")

# Calculate Ratio

train_ratio = len_train_files / (len_train_files + len_test_files + len_val_files)
test_ratio = len_test_files / (len_train_files + len_test_files + len_val_files)
val_ratio = len_val_files / (len_train_files + len_test_files + len_val_files)

print(f"Train Ratio: {train_ratio}")
print(f"Test Ratio: {test_ratio}")
print(f"Val Ratio: {val_ratio}")
```
