```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

IMAGE_PATH = "/Users/lukepitstick/Projects/Data-Science/InfraDrone/datasets/detection/RD2022/test/images/China_Drone_000401.jpg"
```

```python
image = cv2.imread(IMAGE_PATH)
image = np.array(image)
```

```python
x1, y1, x2, y2 = 50, 50, 350, 400

cropped_image = image[
    y1:y2,
    x1:x2,
    :,
]
```

```python
x1, y1, x2, y2 = 100, 100, 200, 200

# Show the crop rectangle on the cropped_image plot
plt.imshow(cropped_image)
plt.scatter([x1, x2, x2, x1, x1], [y1, y1, y2, y2, y1], c='red', s=30)
plt.scatter([x1], [y2], c='blue', s=30)
plt.show()
```
